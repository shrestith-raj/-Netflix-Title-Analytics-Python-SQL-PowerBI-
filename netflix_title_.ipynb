### STEP:5 Handle Missing Values

```python
df['director']=df['director'].fillna('Unknown')
```

```python
df['cast']=df['cast'].fillna('Unknown')
```

```python
df['country']=df['country'].fillna('Unknown')
```

```python
df['date_added']=pd.to_datetime(df['date_added'],errors='coerce')
```

### STEP:6 Check Duplicates

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

### STEP:8 Data Standardization

```python
df['duration_value']=df['duration'].str.extract('(\d+)')
```

```python
df['duration_value'] = pd.to_numeric(df['duration_value'],errors='coerce')
```

```python
df['duration_unit'] = df['duration'].str.extract('(min|Season|Seasons)')
```

```python
df['added_year']=df['date_added'].dt.year
```

```python
df['added_month']=df['date_added'].dt.month
```

```python
df['added_month_name']=df['date_added'].dt.month_name()
```

### STEP:10 EDA 

### EDA:1 Movies vs Shows 

```python
df['type'].value_counts()
```

```python
sns.countplot(data=df, x='type')
plt.title('Movies vs TV Shows')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')
plt.show()
```

### EDA:2 Content Added By Year

```python
content_by_year = df['added_year'].value_counts().sort_index()
content_by_year.plot(kind='line', figsize=(12,5))
plt.title('Netflix Content Added by Year')
plt.xlabel('Year')
plt.ylabel('Number of Titles')
plt.show()
```

### EDA:3 Release Year Distribution

```python
plt.figure(figsize=(12,5))
sns.histplot(df['release_year'],bins=20,kde=True)
plt.title('Distribution of Release Years')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.show()
```

### EDA:4 Top Countries 

```python
country_df = df.assign(
country=df['country'].str.split(',')).explode('country')
country_df['country'] = country_df['country'].str.strip()
```

```python
top_countries = country_df['country'].value_counts().head(10)
top_countries.plot(kind='barh')
plt.title('Top 10 Countries by Netflix Titles')
plt.xlabel('Number of Titles')
plt.ylabel('Country')
plt.show()
```

### EDA:5 Top genres

```python
genre_df = df.assign(
genre=df['listed_in'].str.split(',')).explode('genre')
genre_df['genre'] = genre_df['genre'].str.strip()
```

```python
top_genres = genre_df['genre'].value_counts().head(10)
top_genres.plot(kind='barh')
plt.title('Top 10 Netflix Genres')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')
plt.show()
```

### EDA:6 Ratings Distribution

```python
rating_counts = df['rating'].value_counts()
rating_counts.plot(kind='bar', figsize=(10,5))
plt.title('Netflix Content by Rating')
plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.xticks(rotation=45)
plt.show()
```

### EDA:7 Movie Distribution

```python
movies = df[df['type'] == 'Movie']
```

```python
movies['duration_value'].describe()
```

```python
sns.histplot(movies['duration_value'].dropna(), bins=30)

plt.title('Distribution of Movie Duration')
plt.xlabel('Duration in Minutes')
plt.ylabel('Number of Movies')
plt.show()
```
